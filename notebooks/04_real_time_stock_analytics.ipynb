```python
import requests
import time
import random

# same 20 stocks
symbols = [
    "RELIANCE","TCS","INFY","HDFCBANK","ICICIBANK",
    "SBIN","LT","ITC","HINDUNILVR","KOTAKBANK",
    "AXISBANK","ASIANPAINT","MARUTI","SUNPHARMA",
    "TITAN","ULTRACEMCO","BAJFINANCE","WIPRO",
    "NTPC","POWERGRID"
]

# storage
price_history = {}
volume_history = {}
```

```python
def fetch_nse(symbol):
    url = f"https://www.nseindia.com/api/quote-equity?symbol={symbol}"

    headers = {
        "User-Agent": "Mozilla/5.0",
        "Accept-Language": "en-US,en;q=0.9",
        "Accept-Encoding": "gzip, deflate, br",
        "Referer": "https://www.nseindia.com/",
        "Connection": "keep-alive"
    }

    try:
        session = requests.Session()

        # important (prevents blocking)
        session.get("https://www.nseindia.com", headers=headers)

        response = session.get(url, headers=headers)

        if response.status_code != 200:
            return None

        return response.json()

    except:
        return None
```

```python
def extract_features(data):
    features = {}

    price_info = data.get('priceInfo', {})

    features['price'] = price_info.get('lastPrice')
    features['pChange'] = price_info.get('pChange')

    features['volume'] = data.get('preOpenMarket', {}).get('totalTradedVolume')

    return features
```

```python
for stock in price_history:

    prices = price_history[stock]

    if len(prices) > 1:
        ret = (prices[-1] - prices[-2]) / prices[-2]

        print(f"{stock} Return:", round(ret, 5))
```
